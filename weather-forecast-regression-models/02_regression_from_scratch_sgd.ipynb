```python
French_cities = ['MONTELIMAR', 'PERPIGNAN', 'TOURS']
French_columns = ['DATE', 'MONTH']

for city in French_cities:
    for col in df.columns:
        if col.startswith(city):
            French_columns.append(col)

df_French = df[French_columns]
print(df_French)
```

```python
df_French['DATE'] = pd.to_datetime(df_French['DATE'], format='%Y%m%d')
df_French= df_French[df_French['DATE'].dt.year == 2009]
print(df_French.head())
```

```python
import pandas as pd

normalized_columns = df_French.columns.drop(['DATE', 'MONTH'])

for col in normalized_columns:
    # Calculate min and max for the current column
    min_val = df_French[col].min()
    max_val = df_French[col].max()

    # Apply the normalization formula
    df_French[col] = (df_French[col] - min_val) / (max_val - min_val)

print(df_French.head())
```

#Temprature Tours

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from tqdm import tqdm
#This function is used to create progress bars,
# which are helpful for monitoring the progress of loops, especially during long-running tasks like model training.

city_a, city_b, city_c = 'TOURS', 'MONTELIMAR', 'MONTELIMAR'
X = df_French[[
    f'{city_a}_humidity', f'{city_a}_temp_mean',
    f'{city_b}_humidity', f'{city_b}_temp_mean',
    f'{city_c}_humidity', f'{city_c}_temp_mean'
]].values

y = df_French[f'{city_a}_temp_mean'].values
y = y.reshape(len(y), 1)
#Reshapes the y array into a column vector (a 2D array with one column).
# This is often done to ensure compatibility with matrix operations later in the code.
def split_data(X, y, test_size=0.2, random_state=0):
    np.random.seed(random_state)
    indices = np.random.permutation(len(X))#Generates a shuffled array of indices from 0 to the length of X.
    #This is used to randomly split the data.
    data_test_size = int(X.shape[0] * test_size)#the number of samples to include in the test set
    train_indices = indices[data_test_size:]
    test_indices = indices[:data_test_size]
    X_train = X[train_indices]
    y_train = y[train_indices]
    X_test = X[test_indices]
    y_test = y[test_indices]
    return X_train, y_train, X_test, y_test

class polynomialRegression():

    def __init__(self):
        pass

    def forward(self, X, y, W):
        y_pred = np.dot(W, X)
        loss = np.mean((y_pred - y) ** 2) / 2
        return loss, y_pred

    def updateWeights(self, X, y_pred, y_true, W, alpha, index):
        W -= (alpha * (y_pred - y_true[index]) * X[index]).T
        return W

    def train(self, X, y, X_test, y_test, epochs=1000, alpha=0.001, random_state=0, tolerance=0.001):
        num_rows = X.shape[0]
        num_cols = X.shape[1]
        W = np.random.randn(1, num_cols) / np.sqrt(num_rows)

        train_loss = []
        test_loss = []
        num_epochs = []
        train_indices = [i for i in range(X.shape[0])]

        for j in tqdm(range(epochs), desc="Training", unit="epoch"):  # Use tqdm for progress bar
            cost = 0
            np.random.seed(random_state)
            np.random.shuffle(train_indices)
            for i in train_indices:
                loss, y_pred = self.forward(X[i], y[i], W[0])
                cost += loss
                W[0] = self.updateWeights(X, y_pred, y, W[0], alpha, i)
            train_loss.append(cost)

            test_cost = 0
            for i in range(X_test.shape[0]):
                loss, y_pred = self.forward(X_test[i], y_test[i], W[0])
                test_cost += loss
            test_loss.append(test_cost)

            num_epochs.append(j)

            if cost < tolerance:
                print(f"Training stopped due to tolerance at epoch {j}")
                break

            tqdm.write(f"Epoch {j+1}/{epochs}, Train Loss: {cost:.4f}, Test Loss: {test_cost:.4f}") # Show loss during training

        return W[0], train_loss, test_loss, num_epochs

    def test(self, X_test, y_test, W_trained):
        test_pred = []
        test_loss = []
        test_indices = [i for i in range(X_test.shape[0])]
        for i in test_indices:
            loss, y_test_pred = self.forward(X_test[i], y_test[i], W_trained)
            test_pred.append(y_test_pred)
            test_loss.append(loss)
        return test_pred, test_loss

    def predict(self, W_trained, X_sample):
        prediction = np.dot(W_trained, X_sample)
        return prediction

X_train, y_train, X_test, y_test = split_data(X, y)
model = polynomialRegression()
W_trained, train_loss, test_loss, num_epochs = model.train(X_train, y_train, X_test, y_test, epochs=1000, alpha=0.001, tolerance=0.001)

# Plotting the loss curves
plt.plot(num_epochs, train_loss, label='Train Loss')
plt.plot(num_epochs, test_loss, label='Test Loss')
plt.xlabel('Epochs')
plt.ylabel('Loss')
plt.title('Training and Test Loss Curves')
plt.legend()
plt.show()
```
